## Persistent recovery
Use Save Version → Save & Run All for the authorized long run. Its bounded session exits normally and saves `/kaggle/working/FINet`, training history and `finet_kaggle_recovery_latest.zip` as version outputs. Files only persistence also protects interactive restarts, but is not a substitute for saved outputs. At a later session attach the latest completed version as notebook-output input; the code chooses the highest valid saved epoch and retains its best checkpoint/history. Do not run duplicate trainers. A killed partial epoch is replayed from the previous saved boundary; completed epochs are retained. Free GPU quota is finite and may require waiting for renewal. After A reaches 200 epochs, the same runner uses the original shared full-model initialization for B (lambda 0.1) and proceeds to the existing audited evaluation. Any hardware/package change must be recorded and re-smoked. No quota bypass or paid resources are used.
